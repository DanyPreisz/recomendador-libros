# Recomendador de Libros (Content-Based)

Proyecto de sistema de recomendación listo para **Google Colab**.

**Pregunta:** ¿cómo recomendar libros similares basándonos en género o descripción?

**Método:** TF-IDF + Cosine Similarity

**Stack:** pandas, scikit-learn

Datos de ejemplo.

## 0. Setup

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
print("OK")
```

## 1. Datos de ejemplo

```python
libros = pd.DataFrame({
    "titulo": ["Cien años de soledad", "El amor en los tiempos del cólera", "1984", "Rebelion en la granja", "Don Quijote", "Harry Potter y la piedra filosofal"],
    "descripcion": [
        "realismo mágico familia Macondo soledad",
        "amor pasión espera cólera",
        "distopía totalitarismo vigilancia",
        "alegoría política animales granja",
        "caballería locura molinos aventuras",
        "magia escuela hechiceros amistad"
    ]
})
print(libros)
```

## 2. TF-IDF

```python
tfidf = TfidfVectorizer(stop_words="spanish")
tfidf_matrix = tfidf.fit_transform(libros["descripcion"])
print(tfidf_matrix.shape)
```

## 3. Similitud

```python
similitud = cosine_similarity(tfidf_matrix)
print(similitud)
```

## 4. Función de recomendación

```python
def recomendar(titulo, n=3):
    idx = libros[libros["titulo"] == titulo].index[0]
    scores = list(enumerate(similitud[idx]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)
    top = scores[1:n+1]
    return [libros.iloc[i[0]]["titulo"] for i in top]

print(recomendar("1984"))
```

## 5. Conclusiones

- Content-based con TF-IDF es simple y efectivo para recomendaciones iniciales.
- Se puede extender a canciones (usando género o letras).
- Ideal para combinar con filtrado colaborativo (hybrid).

**Consejo:** usá datasets reales como Goodreads o Spotify.
